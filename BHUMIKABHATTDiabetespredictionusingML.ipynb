```python
# Diabetes Prediction using Machine Learning
# Algorithm: Logistic Regression
# Dataset required: diabetes.csv in the same folder as this file

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Load dataset
try:
    data = pd.read_csv("diabetes.csv")
except FileNotFoundError:
    print("Error: diabetes.csv not found. Put it in the same folder as this program.")
    raise SystemExit

# 2. Check expected columns
features = [
    "Pregnancies", "Glucose", "BloodPressure", "SkinThickness",
    "Insulin", "BMI", "DiabetesPedigreeFunction", "Age"
]
if not all(column in data.columns for column in features + ["Outcome"]):
    print("Dataset columns do not match the standard Pima diabetes dataset.")
    raise SystemExit

X = data[features].copy()
y = data["Outcome"].astype(int)

# Some zero measurements in this dataset may indicate missing values.
zero_as_missing = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
X[zero_as_missing] = X[zero_as_missing].replace(0, float("nan"))

# 3. Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 4. Prepare and train the model
model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=1000, random_state=42))
])
model.fit(X_train, y_train)

# 5. Evaluate model performance on the test set
y_pred = model.predict(X_test)
print("\n===== DIABETES PREDICTION MODEL =====")
print(f"Test accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%")
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=["Outcome 0", "Outcome 1"]))

# 6. Predict for a new example entered by the user
print("\nEnter the person's measurements:")
try:
    values = {
        "Pregnancies": float(input("Number of pregnancies: ")),
        "Glucose": float(input("Glucose level: ")),
        "BloodPressure": float(input("Diastolic blood pressure: ")),
        "SkinThickness": float(input("Skin thickness: ")),
        "Insulin": float(input("Insulin level: ")),
        "BMI": float(input("BMI: ")),
        "DiabetesPedigreeFunction": float(input("Diabetes pedigree function: ")),
        "Age": float(input("Age: "))
    }

    new_person = pd.DataFrame([values], columns=features)
    prediction = model.predict(new_person)[0]
    probability = model.predict_proba(new_person)[0][1] * 100

    print("\n===== PREDICTION RESULT =====")
    if prediction == 1:
        print("Model predicts class 1 (diabetes-positive in this dataset).")
    else:
        print("Model predicts class 0 (diabetes-negative in this dataset).")
    print(f"Model-estimated probability for class 1: {probability:.2f}%")
    print("This is an educational demonstration, not a medical diagnosis.")
    print("Please consult a qualified healthcare professional for testing.")
except ValueError:
    print("Invalid input. Please enter numeric values only.")
```
